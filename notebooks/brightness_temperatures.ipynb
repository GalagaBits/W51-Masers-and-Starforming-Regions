```python
import numpy as np
from astropy import units as u
```

```python
def brightness_temperature(bmaj, bmin, rest_frequency, S):
    bmaj = bmaj*u.arcsec
    bmin = bmin*u.arcsec
    fwhm_to_sigma = 1./(8*np.log(2))**0.5
    beam_area = 2.*np.pi*(bmaj*bmin*fwhm_to_sigma**2)
    freq = rest_frequency*u.GHz
    equiv = u.brightness_temperature(freq)

    temperature = S * (u.Jy/beam_area).to(u.K, equivalencies=equiv)

    print(temperature.value /10**4 )
```
